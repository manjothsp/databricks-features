```python
data1 = [(1,"mouse",300),(2,"keyboard",400),(6,"RAM",0)]
df1 = spark.createDataFrame(data1, ["id","name","price"])
```

```python
from delta import DeltaTable

dt = DeltaTable.forPath(spark, "/Volumes/my_catalog/test/test_volume/products")
```

```python
dt.alias("t").merge(
    source=df1.alias("s"),
    condition="t.id = s.id"
).whenMatchedUpdateAll() \
 .whenNotMatchedInsertAll() \
 .execute()
```
